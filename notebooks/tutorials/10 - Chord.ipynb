```python
### STEREO chord synth, major six nine chord

# create major nine chord, square wave, panned hard left
major_nine = m.chord('maj 9', 'A2', 'square', duration=2).pan('L50').vol(.65)

# create major six chord, saw1 wave, panned hard right
major_six = m.chord('maj 6', 'A2', 'saw1', duration=2).pan('R50')

# add L and R channels together 
major_six_nine = m.add_waves((major_nine, major_six)).vol(.6)

# WAV audio
major_six_nine.bounce()
```

```python
# set tempo
m.set_bpm(180)

# create jazz chord progression
# wide stereo synth

# center mono channel, sine wt_pos = 777
C_F2_lydianchord = m.chord('lydian chord', 'F3', 'sine', duration=1, wt_pos=777)
C_Ab2_lydianchord = m.chord('lydian chord', 'Ab3', 'sine', duration=1, wt_pos=777)
C_D2_minor9 = m.chord('min 9', 'D3', 'sine', duration=1, wt_pos=777)
C_E2_min7b5 = m.chord('min 7 b5', 'E3', 'sine', duration=1, wt_pos=777)
C_channel = m.join_waves((C_F2_lydianchord, C_Ab2_lydianchord, C_D2_minor9, C_E2_min7b5)).vol(.6).LPF(1000)

# left channel (sine wave)
L_F2_lydianchord = m.chord('lydian chord', 'F3', 'sine', duration=1).pan('L50')
L_Ab2_lydianchord = m.chord('lydian chord', 'Ab3', 'sine', duration=1).pan('L50')
L_D2_minor9 = m.chord('min 9', 'D3', 'sine', duration=1).pan('L50')
L_E2_min7b5 = m.chord('min 7 b5', 'E3', 'sine', duration=1).pan('L50')
L_channel = m.join_waves((L_F2_lydianchord, L_Ab2_lydianchord, L_D2_minor9, L_E2_min7b5)).vol(1.2)

# right channel (tri wave)
R_F2_lydianchord = m.chord('lydian chord', 'F3', 'tri', duration=1).pan('R50')
R_Ab2_lydianchord = m.chord('lydian chord', 'Ab3', 'tri', duration=1).pan('R50')
R_D2_minor9 = m.chord('min 9', 'D3', 'tri', duration=1).pan('R50')
R_E2_min7b5 = m.chord('min 7 b5', 'E3', 'tri', duration=1).pan('R50')
R_channel = m.join_waves((R_F2_lydianchord, R_Ab2_lydianchord, R_D2_minor9, R_E2_min7b5))

# join chords together
# progression = m.add_waves((C_channel,L_channel, R_channel)).loop(1).vol(1.1)
progression = m.join_waves((C_channel,L_channel, R_channel)).loop(1).vol(1.1)

progression.bounce()
```
