### Sample

Sample has two inputs:
1. <b>sound_folder</b>- string value that sets the sound folder or category
        - kick
        - snare
        - clap
        - hihat
        - cymbal
        - perc
        - bass
        - fx
        - user
        - archive
2. <b>sample_ID</b> - this selects a particular WAV file in a folder. there are two ways to access samples:
        - index number (integer). example: sample_ID = 0, this will load the first sample in the selected sample_category (indexing starts from 0 not 1)
        - file name (string). example: sample_ID = 'V_RIOT_clap_subtle_room.wav'

### Creating Rhythms

```python
### EXAMPLE 1

kick = m.sample('kick',1).time_edit(1/4)

# create rhythmic pattern increasing in speed
kick_pattern2 = m.join_waves((kick, m.rest(1/4), kick, m.rest(1/6), kick, m.rest(1/8), kick, m.rest(1/12), kick, m.rest(1/16), kick, m.rest(1/24), kick, m.rest(1/32), kick, m.rest(1/48), kick))

kick_pattern2.bounce()
```

```python
### EXAMPLE 2

m.set_bpm(150)

# load samples
kick = m.sample('kick',1).time_edit(1/8)
snare = m.sample('snare',0).time_edit(1/4)

# create rhythmic patterns by using join_waves and rests 
kick_loop = m.join_waves((kick, m.rest(1/4), kick, m.rest(1/8), kick, m.rest(1/4))).loop(2)
snare_loop = m.join_waves((m.rest(1/4) ,snare)).loop(4)

# verify times are the same, if not go back and adjust code
kick_loop.time()
snare_loop.time()

# add kick and snare waves together
mix = m.add_waves((kick_loop, snare_loop))

# WAV audio
mix.bounce()
```

### Create four on floor drum groove

```python
# set tempo
m.set_bpm(130)

# import, time_edit and loop
kick = m.sample('kick',1).time_edit(1/4).loop(16)

clap = m.sample('clap',6).time_edit(1/4)
clap = m.join_waves((m.rest(1/4),clap)).loop(8).vol(.4)

hihat1 = m.sample('hihat',0).time_edit(1/8)
# hihat1 = m.join_waves((m.rest(1/8), hihat1)).loop(16).pan('R15')
hihat1 = m.join_waves((m.rest(1/8), hihat1)).loop(16) #.pan('R15')

mix = m.add_waves((kick,clap,hihat1))

mix.bounce()
# hihat1.bounce()
```

```python
# load user sample
levitate = m.sample('user', 1)

# levitate.bounce()

levitate_01 = m.sample('user', 1)
levitate_02 = m.sample('user', 2)
levitate_03 = m.sample('user', 3)
levitate_04 = m.sample('user', 4)

m.join_waves((levitate_01, m.rest(1/2), levitate_02, m.rest(1/2), levitate_03, m.rest(1/2), levitate_04, m.rest(1/2), )).bounce()

# m.help()
```

```python
m.set_bpm(120)

# loop sample
levitate_full = m.sample('user', 1)

# loop really fast, add 1 measure fade in
levitate_glitch = levitate_full.time_edit(1/16).loop(16).fade(fade_in=1)

# add together with full sample
levitate_final = m.join_waves((levitate_glitch, levitate_full))

# WAV audio
levitate_final.bounce()
```
