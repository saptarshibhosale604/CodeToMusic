# Interactive Wave Visualizer
This animation allows you to change the waveform in real time with sliders.

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from ipywidgets import interact, interactive, fixed, interact_manual
import ipywidgets as widgets
%matplotlib inline

# add scripts folder to path
import sys
# sys.path.insert(1, r'C:\Users\wesle\Desktop\repository\music-code-master\scripts')
# sys.path.insert(1, r'.\Script\CodeToMusic\Music-Code\core')
sys.path.insert(1, r'E:\WORK\PROJECT\git\PersonalAssistant\Script\CodeToMusic\Music-Code\core')

# plot settings
plt.rcParams['figure.figsize'] = (6,4)
plt.rcParams['figure.dpi'] = 150

# import library
from music_code import MusicCode

# initialize
m = MusicCode(120)
```

```python
def animate_wave(f, v, w, c):
    """
    Wave Animation

    Inputs
    f: frequency
    v: volume
    w: wavetable position
    c: fade_length

    """ 
    plt.plot(m.create_wave([f],'sine',1/32, wt_pos=w).fade(fade_in=c,fade_out=c)*v)
    plt.ylim(-4,4)
    
# define value range for each parameter (min, max, interval)
vol_range = (1,8,1)
freq_range = (35,1000,50) 
wt_range = (1,333,32)
fade_range = (1/1024,1/48,1/512)

# call function and create interactive plot
interact(animate_wave, f=freq_range, v=vol_range, w=wt_range, c=fade_range)
```
