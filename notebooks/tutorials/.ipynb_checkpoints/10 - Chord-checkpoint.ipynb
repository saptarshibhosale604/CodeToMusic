# Chord
This allows you to easily generate chords without having to spell them out note by note.

```python
# import library
from music_code import music_code

# initialize
m = music_code.MusicCode(120)
```

```python
# list of all chords you can use in chord method, see below
m.all_chords
```

```python
# major triad
major_triad = m.chord('maj triad', 'A2', 'sine-tri', duration=1).fade(fade_in=1/2,in_curve=7)

major_triad.bounce()
```

```python
# minor triad
minor_triad = m.chord('min triad', 'A2', 'sine-tri', duration=1)

minor_triad.bounce()
```

```python
# diminished triad
dim_triad = m.chord('diminish triad', 'A2', 'sine-tri', duration=1)

dim_triad.bounce()
```

```python
# create augmented triad
aug_triad = m.chord('augment triad', 'A2', 'sine-tri', duration=1)

aug_triad.bounce()
```

```python
# create major 7 chord
maj7_triad = m.chord('maj 7', 'A2', 'sine-tri', duration=1)

maj7_triad.bounce()
```

```python
lydian_chord = m.chord('lydian chord', 'A2', 'sine-tri', duration=1)

lydian_chord.bounce()
```

```python
# minor 11 chord
minor_eleven_chord = m.chord('min 11', 'E2', 'sine-tri', duration=2)

minor_eleven_chord.bounce()
```

```python
### STEREO chord synth, major six nine chord

# create major nine chord, square wave, panned hard left
major_nine = m.chord('maj 9', 'A2', 'square', duration=2).pan('L50').vol(.65)

# create major six chord, saw1 wave, panned hard right
major_six = m.chord('maj 6', 'A2', 'saw1', duration=2).pan('R50')

# add L and R channels together 
major_six_nine = m.add_waves((major_nine, major_six)).vol(.6)

# WAV audio
major_six_nine.bounce()
```

```python
# set tempo
m.set_bpm(180)

# create jazz chord progression
# wide stereo synth

# center mono channel, sine wt_pos = 777
L_F2_lydianchord = m.chord('lydian chord', 'F3', 'sine', duration=1, wt_pos=777)
L_Ab2_lydianchord = m.chord('lydian chord', 'Ab3', 'sine', duration=1, wt_pos=777)
L_D2_minor9 = m.chord('min 9', 'D3', 'sine', duration=1, wt_pos=777)
L_E2_min7b5 = m.chord('min 7 b5', 'E3', 'sine', duration=1, wt_pos=777)
C_channel = m.join_waves((L_F2_lydianchord, L_Ab2_lydianchord, L_D2_minor9, L_E2_min7b5)).vol(.6).LPF(1000)

# left channel (sine wave)
L_F2_lydianchord = m.chord('lydian chord', 'F3', 'sine', duration=1).pan('L50')
L_Ab2_lydianchord = m.chord('lydian chord', 'Ab3', 'sine', duration=1).pan('L50')
L_D2_minor9 = m.chord('min 9', 'D3', 'sine', duration=1).pan('L50')
L_E2_min7b5 = m.chord('min 7 b5', 'E3', 'sine', duration=1).pan('L50')
L_channel = m.join_waves((L_F2_lydianchord, L_Ab2_lydianchord, L_D2_minor9, L_E2_min7b5)).vol(1.2)

# right channel (tri wave)
R_F2_lydianchord = m.chord('lydian chord', 'F3', 'tri', duration=1).pan('R50')
R_Ab2_lydianchord = m.chord('lydian chord', 'Ab3', 'tri', duration=1).pan('R50')
R_D2_minor9 = m.chord('min 9', 'D3', 'tri', duration=1).pan('R50')
R_E2_min7b5 = m.chord('min 7 b5', 'E3', 'tri', duration=1).pan('R50')
R_channel = m.join_waves((R_F2_lydianchord, R_Ab2_lydianchord, R_D2_minor9, R_E2_min7b5))

# join chords together
progression = m.add_waves((C_channel,L_channel, R_channel)).loop(2).vol(1.1)

progression.bounce()
```
