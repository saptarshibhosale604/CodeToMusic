```python
# center mono
wave_mono = m.create_wave(['C3'], 'tri', 1/2).pan('C')

wave_mono.bounce()
```

```python
# pan right
wave_right = m.create_wave(['C3'], 'tri', 1/2).pan('R50')

wave_right.bounce()
```

```python
# pan left
wave_left = m.create_wave(['C3'], 'tri', 1/2).pan('L50')

wave_left.bounce()
```

```python
# pan right (1-50 range)
wave_right = m.create_wave(['C3'], 'saw1', 1/4).pan('R25')

wave_right.bounce()
```

```python
# pan left (1-50 range)
wave_left = m.create_wave(['C3'], 'saw1', 1/4).pan('L33')

wave_left.bounce()
```

```python
# pan left (1-50 range)
wave_left = m.create_wave(['C3'], 'saw1', 1/4).pan('L5')

wave_left.bounce()
```

```python
# join notes into melody
melody = m.join_waves((wave_left, wave_right, wave_left, wave_right))

# check melody waveform duration
melody.time()

melody.loop(3)

# bounce audio
melody.bounce()

```
