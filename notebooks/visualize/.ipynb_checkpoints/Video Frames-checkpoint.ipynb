```python
import os

master_waveform = np.zeros(1)

# notevalues = list(m.freq_note_table_sharp.values())
# notevalues = list(m.freq_note_table_sharp.values())[:50]
notevalues = list(m.freq_note_table_sharp.values())[:20]
# notevalues = list(m.freq_note_table_sharp.values())

freqvalues = np.arange(5, 5000, 25).astype(float)
fade_size = np.linspace(0.031250, 0.0078125, len(notevalues))

output_dir = r"E:\WORK\PROJECT\git\PersonalAssistant\Script\CodeToMusic\Music-Code\notebooks\visualize\Output"
os.makedirs(output_dir, exist_ok=True)  # Ensures the folder exists

iteration_num = 1
wavetable_position = 1

def CreateImages():
    global iteration_num
    global wavetable_position
    for note in notevalues:
        file_name = str(iteration_num) + '.png'
        png_file_path = os.path.join(output_dir, file_name)
    
        # Set up plots
        fig = plt.figure(dpi=1200)
        ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
        fig.patch.set_visible(False)
        ax.axis('off')
    
        # Create waveform
        waveform = m.create_wave([note], 'square', 0.03125, wt_pos=1).fade(fade_in=0.03125, fade_out=0.03125)
    
        # Plot on axis
        ax.plot(waveform)
    
        # Export HD image (savefig handles the file writing entirely)
        fig.savefig(png_file_path, transparent=True)
        
        # Close the figure to free up memory (crucial when generating lots of plots)
        plt.close(fig)
    
        iteration_num += 1
        wavetable_position += 2

CreateImages()

len(notevalues)
# notevalues
```
